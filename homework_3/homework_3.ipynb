# prep

```python
# feats_categ = df_0.dtypes[df_0.dtypes == 'str'].index.tolist()
# feats_numer = df_0.dtypes[(df_0.dtypes != 'str')].index.tolist()
feats_categ = ['lead_source', 'industry', 'employment_status', 'location']
feats_numer = ['annual_income', 'number_of_courses_viewed', 'interaction_count', 'lead_score']

for f in feats_categ:
    print(f"{f} ({len(df_0[f].unique())}): {df_0[f].unique().tolist()}")
    
df_1 = df_0.copy()

# clean
for f in feats_categ:
    df_1[f] = df_1[f].str.lower().str.replace(' ', '_').str.replace('-', '_')
    df_1[f] = df_1[f].fillna('NA')
for f in feats_numer:
    df_1[f] = df_1[f].fillna(0)    
```

```python
df_1.info()
```

# Q1

```python
# What is the most frequent observation (mode) for the column industry?
# Answer: technology at 1,173 freq

print(df_1[feats_categ].describe())
```

# Q2

```python
# What are the two features that have the biggest correlation?
# Answer: interaction_count and lead_score

corr_matrix = df_1[feats_numer].corr()
sns.heatmap(corr_matrix, annot=True)
plt.show()
```

```python
from sklearn.model_selection import train_test_split

df_full_train, df_test = train_test_split(df_1, test_size=0.2, random_state=42)
df_train, df_val = train_test_split(df_full_train, test_size=0.25, random_state=42)

len(df_full_train), len(df_test), len(df_val), len(df_train)

y_full_train = df_full_train.converted.values
y_test = df_test.converted.values
y_val = df_val.converted.values
y_train = df_train.converted.values
```

# Q3

```python
# Which of these variables has the biggest mutual information score?
# Answer: lead_sources at 0.03

from sklearn.metrics import mutual_info_score

for f in feats_categ:
    mutual_score = mutual_info_score(y_train, df_train[f])
    print(f"{f}: {round(mutual_score, 4)}")
```

# Q4

```python
# Calculate the accuracy on the validation dataset and round it to 2 decimal digits.
# What accuracy did you get?
# Answer: 0.65

from sklearn.feature_extraction import DictVectorizer
from sklearn.linear_model import LogisticRegression

feats_all = feats_categ + feats_numer

dv = DictVectorizer(sparse=False)

train_dict = df_train[feats_all].to_dict(orient='records')
X_train = dv.fit_transform(train_dict)

val_dict = df_val[feats_all].to_dict(orient='records')
X_val = dv.fit_transform(val_dict)

model = LogisticRegression(solver='liblinear', C=1.0, max_iter=1000, random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict_proba(X_val)[:,1]
convert_decision = (y_pred >= 0.5)

print((y_val == convert_decision).mean().round(4))
```

# Q5

```python
# For each feature, calculate the difference between the original accuracy and the accuracy without the feature.
# Which of following feature has the smallest difference?
# Answer: among the choices, 'lead_source'

for f in feats_all:
    feats = [x for x in feats_all if x != f]
    
    dv = DictVectorizer(sparse=False)

    train_dict_2 = df_train[feats].to_dict(orient='records')
    X_train_2 = dv.fit_transform(train_dict_2)

    val_dict_2 = df_val[feats].to_dict(orient='records')
    X_val_2 = dv.fit_transform(val_dict_2)

    model = LogisticRegression(solver='liblinear', C=1.0, max_iter=1000, random_state=42)
    model.fit(X_train_2, y_train)

    y_pred_2 = model.predict_proba(X_val_2)[:,1]
    convert_decision_2 = (y_pred_2 >= 0.5)

    print(f"ex-{f}: {(y_val == convert_decision_2).mean().round(4)} ({round(abs(convert_decision.mean() - convert_decision_2.mean()), 10)})")
```

# Q6

```python
# Let's try the following values of the parameter C: [0.000001, 0.00001, 0.0001, 0.001].
# Calculate the accuracy on the validation dataset and round it to 3 decimal digits.
# Which of these C leads to the best accuracy on the validation set?
# Answer: 0.001 at 64.4% accuracy

for c in [0.000001, 0.00001, 0.0001, 0.001, 0.01, 0.1, 1]:
    dv = DictVectorizer(sparse=False)

    train_dict_3 = df_train[feats].to_dict(orient='records')
    X_train_3 = dv.fit_transform(train_dict_3)

    val_dict_3 = df_val[feats].to_dict(orient='records')
    X_val_3 = dv.fit_transform(val_dict_3)

    model = LogisticRegression(solver='liblinear', C=c, max_iter=1000, random_state=42)
    model.fit(X_train_3, y_train)

    y_pred_3 = model.predict_proba(X_val_3)[:,1]
    convert_decision_3 = (y_pred_3 >= 0.5)

    print(f"c={c:.6f}: {(y_val == convert_decision_3).mean().round(4)}")
```
