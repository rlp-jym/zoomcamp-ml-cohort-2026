```python
df_hp_median = df.horsepower.median()
df_hp_mode = df.horsepower.mode()[0]

df_copy = df.copy()
df_copy['horsepower'] = df_copy.horsepower.fillna(df_hp_mode)
df_copy

df_copy_hp_median = df_copy.horsepower.median()
print("Q6. Decreased")
print(f"    Median hp:   {df_hp_median}")
print(f"    Adj. median: {df_copy_hp_median}")
```

```python
X = df[df.origin == 'Asia'][['vehicle_weight', 'model_year']].head(7).values
XTX = X.T

XTX_inv = np.linalg.inv(XTX @ X) # need to google '@ X'; not in notes

y = [1100, 1300, 800, 900, 1000, 1100, 1200]

w = XTX_inv.dot(XTX) * y

print(f"Q7. Whatever I just did: {round(w.sum(), 3)}")
```
